# Relación de anomalías mensuales de temperatura superficial del mar con la captura por lance de atún aleta amarilla (*Thunnus albacares*) en la zona del Domo Térmico de Costa Rica

**Autores:** Jesus, Silvia y Barbara

## Introducción

El Domo Térmico de Costa Rica (DTCR) es un sistema de afloramiento (upwelling) oceánico ubicado en el océano Pacífico Oriental Tropical, aproximadamente a unos 300 km frente al Golfo de Papagayo, entre Costa Rica y Nicaragua. Se caracteriza por la elevación de la termoclina, que favorece el ascenso de aguas subsuperficiales frías y ricas en nutrientes, propiciando condiciones de alta alta productividad biológica (Fiedler, 2002). Este sistema presenta un ciclo estacional, asociado al forzamiento del viento, modificando su posición, extensión y profundidad a lo largo del año (Fiedler, 2002). 

Considerando la influencia de estos procesos oceanográficos sobre las condiciones ambientales y la distribución de los recursos pesqueros, en este  trabajo se analizará la variabilidad de la temperatura superficial del mar (TSM) en la zona de influencia del DTCR y su posible relación con el atún aleta amarilla (*Thunnus albacares*).


Cita: Fiedler, P. C. (2002). The annual cycle and biological effects of the Costa Rica Dome. *Deep-Sea Research Part I: Oceanographic Research Papers*, 49(2), 321–338. https://doi.org/10.1016/S0967-0637(01)00057-7
